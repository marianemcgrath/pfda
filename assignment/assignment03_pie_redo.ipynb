```python
# The first step here is to load the necessary libraries to run the full task
# We'll import Pandas the data and Matplotlib for plotting

import pandas as pd
import matplotlib.pyplot as plt


# Then we'll load and read the data from the URL provided
url = "https://drive.google.com/uc?id=1AWPf-pJodJKeHsARQK_RHiNsE8fjPCVK&export=download"
data = pd.read_csv(url)

print("Libraries loaded and data read successfully!")
```

```python
# Before we start the analysis, let's have a look at the data we are handling
display(data.head())

# Now, we'll go through the data, and extract the email domains, or anything after @ (which is what we want) and put it into a new column
data['domain'] = data['Email'].str.split('@').str[1]

# Source: https://pandas.pydata.org/docs/reference/api/pandas.Series.str.split.html
# Source: https://www.w3schools.com/python/ref_string_split.asp
# Source: https://www.geeksforgeeks.org/python/python-string-split/
```

```python
# Then, we calculate the percentage for each domain
domain_counts = data['domain'].value_counts()
domain_percentages = (domain_counts / len(data)) * 100

# Finally, we'll print the domain percentages
print("\nDomain Percentages:")
print(domain_percentages)

# Source: https://stackoverflow.com/questions/56294597/how-to-create-a-pie-chart-from-csv-file-using-python
# Source: https://www.analyticsvidhya.com/blog/2024/02/pie-chart-matplotlib
```

```python
# First, let's calculate the domain counts and percentages
domain_counts = data['domain'].value_counts()
domain_percentages = (domain_counts / len(data)) * 100

# And now for the big reveal, we'll plot the pie chart with some pastel colours and black edges to enhance visualisation
plt.figure(figsize=(14, 8))
plt.pie(domain_percentages, labels=domain_percentages.index, autopct='%1.1f%%', colors=['peachpuff','rosybrown','gray'],
        wedgeprops={'edgecolor': 'black'}, shadow={'ox': -0.04, 'edgecolor': 'none', 'shade': 0.9},startangle=100,
        textprops={'fontsize': 14, 'fontweight': 'bold'}, explode=[0.05]*len(domain_percentages))

# We'll make the pie chart have 3D effect by adding a shadow
plt.gca().set_aspect('equal')  # Equal aspect ratio ensures that pie is drawn as a circle
plt.title("Distribution of Email Domains")
plt.show()

# Source: https://matplotlib.org/stable/gallery/pie_and_polar_charts/pie_features.html
# Source: https://stackoverflow.com/questions/56152451/how-to-create-a-pie-chart-using-matplotlib-from-csv
# Source: https://www.geeksforgeeks.org/data-science/plot-a-pie-chart-in-python-using-matplotlib
# Source: https://www.w3schools.com/python/matplotlib_pie_charts.asp
```
