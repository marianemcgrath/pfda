### Task 1: Display a plot of projected birth rates

```python
# Import necessary Libraries

import requests as reqs
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json as js

# Load the dataset from URL
url = "https://github.com/andrewbeattycourseware/PFDA-courseware/blob/main/code/data/projectedbirths-cso.csv"

data = pd.read_csv(url + "?raw=true")

# Display the first 5 rows of the dataset
display (data)

# Source: https://www.youtube.com/watch?v=7vakq_2lHJM (Birth Rate Plot per Country / Year)
```

```python
# Plottting the data
data.info()
```

####  Plotting the data

```python
plot = sns.lineplot(data=data, x= "Year", y= "VALUE", hue= "VALUE", marker= "o")
plt.title("Projected Birth dates in Ireland")
plt.xlabel("Year")
plt.ylabel("Number of Births")
plt.show()
```
